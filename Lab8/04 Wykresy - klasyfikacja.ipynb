# Wizualna ocena modeli modeli klasyfikacyjnych

Ocena modeli klasyfikacyjnych za pomocą **interpretacji graficznej** stanowi ważny etap analizy jakości modelu. Uzupełnia standardowe metryki liczbowe, takie jak accuracy, precision, recall, F1-score czy ROC AUC, o możliwość lepszego zrozumienia zachowania modelu w różnych scenariuszach decyzyjnych.

Wizualizacja wyników klasyfikacji pozwala:

* ocenić kompromis między wykrywaniem klasy pozytywnej a liczbą fałszywych alarmów,
* przeanalizować wpływ zmiany progu decyzyjnego na jakość klasyfikacji,
* sprawdzić strukturę poprawnych i błędnych predykcji,
* ocenić skuteczność modelu przy niezrównoważonych klasach,
* sprawdzić, czy przewidywane prawdopodobieństwa są dobrze skalibrowane,
* ocenić, czy model dobrze separuje obserwacje należące do różnych klas.

Graficzna interpretacja pozwala lepiej zrozumieć mocne i słabe strony modelu. Może również wskazać konieczność jego poprawy, na przykład przez dostrojenie progu decyzyjnego, zastosowanie metod równoważenia klas, zmianę metryki optymalizacyjnej albo kalibrację przewidywanych prawdopodobieństw.

---

## Najczęściej stosowane wykresy do oceny modeli klasyfikacyjnych

**ROC Curve** - **Receiver Operating Characteristic Curve**

Wykres ROC pokazuje zależność między czułością, czyli True Positive Rate, a odsetkiem fałszywie pozytywnych klasyfikacji, czyli False Positive Rate, dla różnych wartości progu decyzyjnego.

Pozwala ocenić zdolność modelu do odróżniania klasy pozytywnej od negatywnej. Pole pod krzywą ROC, czyli ROC AUC, jest miarą ogólnej zdolności klasyfikatora do separacji klas.

---

**PR Curve - Precision-Recall Curve**

Wykres Precision-Recall przedstawia zależność między precyzją a czułością dla różnych progów decyzyjnych.

Jest szczególnie przydatny w przypadku niezrównoważonych klas, zwłaszcza gdy klasa pozytywna jest rzadka. W takich sytuacjach Precision-Recall Curve może być bardziej informacyjna niż ROC Curve, ponieważ silniej koncentruje się na jakości wykrywania klasy pozytywnej.

---

**DET Curve – Detection Error Tradeoff Curve** - Krzywa kompromisu między błędami detekcji

DET Curve służy do oceny klasyfikatora binarnego, podobnie jak krzywa ROC, ale prezentuje dane w sposób, który bardziej uwidacznia różnice w jakości klasyfikacji, zwłaszcza przy niskich poziomach błędów.

---

**Confusion Matrix** - **Macierz pomyłek**

Macierz pomyłek pokazuje liczbę poprawnych i błędnych klasyfikacji dla poszczególnych klas.

Dla klasyfikacji binarnej przedstawia liczbę przypadków:

* true positive,
* true negative,
* false positive,
* false negative.

Na podstawie macierzy pomyłek można obliczyć wiele podstawowych metryk, takich jak accuracy, precision, recall, specificity oraz F1-score. Jest to jedno z najważniejszych narzędzi do oceny, jakie typy błędów popełnia model.

---

**Histogram przewidywanych prawdopodobieństw**

**Histogram of Predicted Probabilities**

Histogram przewidywanych prawdopodobieństw pokazuje rozkład wartości prawdopodobieństw przypisanych przez model do danej klasy, najczęściej klasy pozytywnej.

Pozwala ocenić, czy model dobrze rozdziela obserwacje należące do różnych klas. Jeżeli prawdopodobieństwa dla klasy pozytywnej i negatywnej wyraźnie się rozdzielają, model dobrze separuje klasy. Jeżeli rozkłady silnie się nakładają, oznacza to, że model ma trudności z odróżnianiem klas.

Wykres ten może również wskazać, czy model generuje predykcje bardzo pewne, bliskie 0 lub 1, czy raczej niepewne, skupione wokół wartości pośrednich.

---

**Calibration Curve**

**Reliability Diagram**

Krzywa kalibracji porównuje przewidywane prawdopodobieństwa z rzeczywistą częstością występowania klasy pozytywnej.

Dobrze skalibrowany model powinien przewidywać prawdopodobieństwa zgodne z rzeczywistością. Na przykład wśród obserwacji, którym model przypisuje prawdopodobieństwo około 0,8, klasa pozytywna powinna występować w przybliżeniu w 80% przypadków.

Calibration Curve jest szczególnie ważna wtedy, gdy interpretujemy predykcje modelu jako prawdopodobieństwa, a nie tylko jako decyzję klasyfikacyjną.

---

**Lift Chart / Gain Chart**

Wykresy Lift Chart i Gain Chart pokazują, jak skutecznie model identyfikuje obserwacje należące do klasy pozytywnej w porównaniu z wyborem losowym.

Są często stosowane w zastosowaniach biznesowych i marketingowych, na przykład przy ocenie modeli wskazujących klientów najbardziej skłonnych do zakupu produktu, rezygnacji z usługi lub odpowiedzi na kampanię.

Wykresy te są szczególnie przydatne wtedy, gdy model służy do uszeregowania obserwacji według przewidywanego prawdopodobieństwa, a nie tylko do przypisania klasy.

---

**CAP Curve**

**Cumulative Accuracy Profile Curve**

CAP Curve pokazuje, jaki odsetek obserwacji należących do klasy pozytywnej można wykryć, analizując określony procent obserwacji uporządkowanych według przewidywanego prawdopodobieństwa.

Podobnie jak Lift Chart i Gain Chart, CAP Curve jest szczególnie przydatna w problemach rankingowych, gdzie istotne jest wskazanie najbardziej prawdopodobnych przypadków klasy pozytywnej.

---

**Podsumowanie**

Graficzna analiza modeli klasyfikacyjnych **nie zastępuje metryk liczbowych**, lecz pozwala lepiej zrozumieć sposób działania modelu.

Wykresy takie jak ROC Curve, Precision-Recall Curve, Confusion Matrix, histogram przewidywanych prawdopodobieństw oraz Calibration Curve pomagają ocenić nie tylko ogólną skuteczność klasyfikatora, ale również rodzaje popełnianych błędów, jakość separacji klas, wpływ progu decyzyjnego oraz wiarygodność przewidywanych prawdopodobieństw.

W praktyce najlepiej stosować kilka wykresów jednocześnie, ponieważ każdy z nich pokazuje inny aspekt działania modelu.

Ocena modeli klasyfikacyjnych za pomocą interpretacji graficznej to istotny etap analizy, który umożliwia nie tylko ilościową ocenę skuteczności predykcji (np. poprzez miary takie jak dokładność czy AUC), ale także zrozumienie zachowania modelu w różnych scenariuszach decyzyjnych. Wykresy te pomagają:

* dostrzec kompromis między błędami fałszywie pozytywnymi a fałszywie negatywnymi (np. ROC, Precision-Recall),
* ocenić jakość i równowagę klasyfikacji (Confusion Matrix),
* sprawdzić, czy prawdopodobieństwa przewidywane przez model są realistyczne (Calibration Curve),
* ocenić rozdzielczość między klasami (Histogram przewidywanych prawdopodobieństw).

Graficzna interpretacja pozwala lepiej zrozumieć mocne i słabe strony modelu oraz może wskazać konieczność jego poprawy, np. poprzez dostrojenie progu decyzyjnego, zastosowanie metod balansowania klas lub kalibrację predykcji. W praktyce stosowanie wielu wykresów jednocześnie daje najbardziej kompletny obraz działania modelu.

## Tworzenie modelu

**Wczytanie danych**:

```python
from sklearn.datasets import fetch_openml
```

## ROC Curve - Receiver Operating Characteristic Curve

Wykres ROC służy do oceny jakości modelu klasyfikacyjnego, **szczególnie w klasyfikacji binarnej**. Pokazuje, jak model zachowuje się przy różnych progach decyzyjnych.

---

**Osie wykresu ROC**:

*Oś X* – **False Positive Rate (FPR)**

$$
\text{FPR} = \frac{\text{FP}}{\text{FP} + \text{TN}}
$$

gdzie:

* **FP** – liczba fałszywie pozytywnych przypadków
* **TN** – liczba prawdziwie negatywnych przypadków

Warto zauważyć, że **specificity (True Negative Rate)** to:

$$
\text{Specificity} = \frac{\text{TN}}{\text{TN} + \text{FP}} = 1 - \text{FPR}
$$

* Specificity opisuje, jak dobrze model rozpoznaje **klasę negatywną**.
* Im wyższa **specificity**, tym niższy **FPR** – są swoim **lustrzanym odbiciem**.

*Oś Y* – **True Positive Rate (TPR), czyli czułość (Recall)**

$$
\text{TPR} = \frac{\text{TP}}{\text{TP} + \text{FN}}
$$

gdzie:

* **TP** – liczba prawdziwie pozytywnych przypadków
* **FN** – liczba fałszywie negatywnych przypadków

---

**Tworzenie krzywej ROC**:

Dla każdej wartości progu $t \in [0, 1]$:

1. Tworzona jest klasyfikacja: $y_{\text{pred}} = (y_{\text{proba}} \geq t)$
2. Obliczane są TPR i FPR
3. Punkt $(\text{FPR}(t), \text{TPR}(t))$ dodawany jest do wykresu

Z wielu takich punktów powstaje **ciągła krzywa ROC**.

---

**Interpretacja**:

* Punkt **(0,1)** oznacza idealny model: **0% fałszywie pozytywnych** i **100% prawdziwie pozytywnych**.
* Linia przekątna $(0,0) \rightarrow (1,1)$ reprezentuje **model losowy**.
* Im bliżej krzywej do lewego górnego rogu, tym lepszy model.

---

**AUC – Area Under Curve**:

* **AUC** to pole powierzchni pod krzywą ROC.
* Im wyższa wartość AUC, tym lepszy model.

Typowe wartości:

* **1.0** – klasyfikator idealny
* **0.5** – klasyfikator losowy
* **< 0.5** – klasyfikator gorszy niż losowy (np. zamienia klasy)

---

**Zalety wykresu ROC**:

* **Nie zależy od jednego progu** – pokazuje skuteczność dla wszystkich progów decyzyjnych.
* **Dobrze sprawdza się przy zrównoważonych klasach**. W przypadku silnie niezrównoważonych klas lepszym wyborem jest często **Precision-Recall Curve**.

---

```python
from sklearn.metrics import roc_curve, auc
```

```python
# Warto sprawdzić model i jego klasy, ROC domyślnie jest dla klasy = 1 jako pozytywnej, u Nas bad=1
```

```python
fpr, tpr, thresholds = roc_curve(y_test, y_proba)
roc_auc = auc(fpr, tpr)

fig, ax = plt.subplots()
ax.plot(fpr, tpr, label=f"ROC curve (AUC = {roc_auc:.4f})")
ax.plot([0, 1], [0, 1], linestyle="--", color="gray")
ax.grid(color="gray", linestyle="--", linewidth=0.7, alpha=0.6)
ax.set_xlabel("False Positive Rate (FPR)")
ax.set_ylabel("True Positive Rate (TPR)")
ax.set_title("ROC Curve")
ax.legend()

# Dodanie wartości progów co n-ty punkt, UWAGA: standardowo progów się nie nanosi na krzywa ROC
step = max(1, len(thresholds) // 20)
for i in range(0, len(thresholds), step):
    ax.annotate(f"{thresholds[i]:.2f}", (fpr[i], tpr[i]), textcoords="offset points", xytext=(0, -10), ha='center', fontsize=8)

plt.show()
```

**Interpretacja krzywej ROC** 

* **AUC = 0.6824** oznacza, że model ma umiarkowaną zdolność rozróżniania między klasą pozytywną a negatywną. 68% szans, że model przypisze inną wartość predykcji losowo wybranemu przypadkowi pozytywnemu niż losowo wybranemu przypadkowi negatywnemu.

* Krzywa ROC pokazuje **typowy kompromis**: wyższy TPR = wyższy FPR.
* **Dla progów 0.3–0.35** model osiąga już znaczącą czułość (`TPR ≈ 0.5–0.6`), ale z umiarkowanym FPR (`0.25–0.3`).
* Najlepsze progi zależą od priorytetu:

  * Jeśli ważniejsze jest **uniknięcie błędów fałszywie pozytywnych** → wybierz **wyższy próg** (mniejszy FPR, kosztem TPR).
  * Jeśli ważniejsze jest **wykrycie wszystkich pozytywnych** → wybierz **niższy próg** (większy TPR, kosztem FPR).

Poniżej znajduje się kod, który umożliwia dodanie dowolnej liczby modeli klasyfikacyjnych do listy słownikowej i automatyczne wygenerowanie wykresu ROC z porównaniem ich skuteczności (AUC):

```python
from sklearn.linear_model import LogisticRegression
from sklearn.neighbors import KNeighborsClassifier
from sklearn.ensemble import RandomForestClassifier
```

```python
# Lista modeli
models = {
    "Logistic Regression": LogisticRegression(max_iter=10000),
    "KNN (k=3)": KNeighborsClassifier(n_neighbors=3),
    "Random Forest": RandomForestClassifier(random_state=42)
}

# Wykres ROC
fig, ax = plt.subplots()

for name, model in models.items():
    model.fit(X_train, y_train)
    y_proba = model.predict_proba(X_test)[:, 1]
    fpr, tpr, _ = roc_curve(y_test, y_proba) # <=== tu jest funkcja roc_curve()
    roc_auc = auc(fpr, tpr)
    ax.plot(fpr, tpr, label=f"{name} (AUC = {roc_auc:.4f})")

ax.plot([0, 1], [0, 1], linestyle="--", color="gray")
ax.set_xlabel("False Positive Rate")
ax.set_ylabel("True Positive Rate")
ax.set_title("ROC Curve Comparison")
ax.legend()
plt.show()
```

Do utworzenia krzywej ROC w bibliotece scikit-learn można wykorzystać klasę RocCurveDisplay, która udostępnia kilka metod tworzenia wykresu w zależności od tego, czy pracujemy bezpośrednio z modelem, gotowymi predykcjami, czy wynikami walidacji krzyżowej.


* **RocCurveDisplay.from_estimator** – tworzy krzywą ROC na podstawie wytrenowanego modelu oraz danych wejściowych, samodzielnie obliczając predykcje lub wyniki decyzyjne.
* **RocCurveDisplay.from_predictions** – tworzy krzywą ROC na podstawie podanych wartości rzeczywistych oraz wcześniej obliczonych wyników predykcji, np. prawdopodobieństw klasy pozytywnej.
* **RocCurveDisplay.from_cv_results** – tworzy zestaw krzywych ROC dla wielu podziałów walidacji krzyżowej na podstawie wyników uzyskanych z cross-validation.

```python
import matplotlib.pyplot as plt

from sklearn.linear_model import LogisticRegression
from sklearn.metrics import RocCurveDisplay
from sklearn.model_selection import cross_validate

# Model
model = LogisticRegression(max_iter=10000)
model.fit(X_train, y_train)

# Predykcje
y_proba = model.predict_proba(X_test)[:, 1]
y_pred = model.predict(X_test)
```

**ROC z użyciem from_estimator**

Funkcja sama oblicza prawdopodobieństwa lub wartości decyzyjne modelu.

```python
RocCurveDisplay.from_estimator(
    model,
    X_test,
    y_test,
    pos_label=1,
    plot_chance_level=True,
    name="Logistic Regression"
)

plt.title("ROC Curve - from_estimator")
plt.grid(True, linestyle="--", alpha=0.6)
plt.show()
```

**ROC z użyciem from_predictions**

Tutaj wykorzystujesz wcześniej obliczone prawdopodobieństwa klasy pozytywnej.

```python
RocCurveDisplay.from_predictions(
    y_test,
    y_proba,
    pos_label=1,
    plot_chance_level=True,
    name="Logistic Regression"
)

plt.title("ROC Curve - from_predictions")
plt.grid(True, linestyle="--", alpha=0.6)
plt.show()
```

**ROC z użyciem from_cv_results**

Ten wariant służy do pokazania krzywych ROC dla kilku podziałów walidacji krzyżowej. W scikit-learn metoda from_cv_results jest dostępna od wersji 1.7 i wymaga wyników cross_validate z return_estimator=True oraz return_indices=True.

```python
cv_results = cross_validate(
    model,
    X_train,
    y_train,
    cv=5,
    return_estimator=True,
    return_indices=True
)

RocCurveDisplay.from_cv_results(
    cv_results = cv_results,
    X = X_train,
    y = y_train,
    pos_label=1,
    name="Logistic Regression"
)

plt.title("ROC Curve - cross-validation")
plt.grid(True, linestyle="--", alpha=0.6)
plt.show()
```

## PR Curve - Precision-Recall Curve

Wykres Precision-Recall (PR) służy do oceny jakości modelu klasyfikacyjnego, **szczególnie przy niezrównoważonych zbiorach danych**. Pokazuje, jak model radzi sobie z wykrywaniem klasy pozytywnej przy różnych progach decyzyjnych.

---

**Osie wykresu PR**:

*Oś X* – **Recall (TPR, czułość)**

$$
\text{Recall} = \frac{\text{TP}}{\text{TP} + \text{FN}}
$$

gdzie:

* **TP** – liczba prawdziwie pozytywnych przypadków
* **FN** – liczba fałszywie negatywnych przypadków

Opisuje, jaki **odsetek rzeczywistych przykładów pozytywnych został poprawnie wykryty** przez model.

*Oś Y* – **Precision (precyzja)**

$$
\text{Precision} = \frac{\text{TP}}{\text{TP} + \text{FP}}
$$

gdzie:

* **TP** – liczba prawdziwie pozytywnych przypadków
* **FP** – liczba fałszywie pozytywnych przypadków

Opisuje, **jaki odsetek zaklasyfikowanych jako pozytywne** rzeczywiście był pozytywny.

---

**Tworzenie krzywej PR**:

Dla każdej wartości progu $t \in [0, 1]$:

1. Tworzona jest klasyfikacja: $y_{\text{pred}} = (y_{\text{proba}} \geq t)$
2. Obliczane są wartości Precision i Recall
3. Punkt $(\text{Recall}(t), \text{Precision}(t))$ dodawany jest do wykresu

Z tych punktów powstaje **krzywa Precision-Recall**, której kształt zależy od proporcji klas i skuteczności modelu.

---

**Interpretacja**:

* Im większa **precyzja przy danym recall**, tym lepszy model.
* Idealny model: punkt **(1, 1)** – 100% trafień, 0% fałszywych alarmów.
* W przeciwieństwie do ROC, tutaj nawet dobra wartość recall może być bezwartościowa, jeśli precision jest niska (i odwrotnie).
* Linia bazowa to wartość **prevalence** (udział klasy pozytywnej) — model losowy osiąga precision równą częstości klasy pozytywnej.

---

**AUC dla PR – Average Precision (AP)**:

Miara **AUC w kontekście PR** nazywa się **Average Precision** (AP). Jest to uśredniona precyzja ważona przyrostami recall:

$$
\text{AP} = \sum_n (\text{Recall}_n - \text{Recall}_{n-1}) \cdot \text{Precision}_n
$$

AP można traktować jako pole powierzchni pod wykresem PR. Im wyższe AP, tym lepszy model.

---

**Zalety wykresu Precision-Recall**:

* **Lepiej niż ROC** pokazuje działanie modelu przy **silnie niezrównoważonych klasach**.
* Bardziej rygorystyczny niż ROC — ukazuje skuteczność tylko w odniesieniu do **klasy pozytywnej**, bez „maskowania” wynikami klasy negatywnej.
* Ujawnia kompromis: **więcej trafień (recall) może oznaczać mniej trafnych wyników (precision)**.

---

```python
from sklearn.metrics import precision_recall_curve, average_precision_score
```

```python
precision, recall, thresholds = precision_recall_curve(y_test, y_proba)
ap = average_precision_score(y_test, y_proba)

fig, ax = plt.subplots()

ax.plot(
    recall,
    precision,
    drawstyle="steps-post", # <==== Dla Precision-Recall bardziej standardowa jest wersja schodkowa
    label=f"PR curve (AP = {ap:.4f})"
)

ax.grid(color="gray", linestyle="--", linewidth=0.7, alpha=0.6)
ax.set_xlabel("Recall")
ax.set_ylabel("Precision")
ax.set_title("Precision-Recall Curve")
ax.legend()

# Dodanie wartości progów co n-ty punkt
step = max(1, len(thresholds) // 20)

for i in range(0, len(thresholds), step):
    ax.annotate(
        f"{thresholds[i]:.2f}",
        (recall[i], precision[i]),
        textcoords="offset points",
        xytext=(0, -10),
        ha="center",
        fontsize=8
    )

plt.show()
```

**Interpretacja**:

* Wartość AP = 0.4942 (Average Precision) oznacza, że model osiąga średnią precyzję około 49,42% przy wykrywaniu klasy pozytywnej w całym zakresie progów decyzyjnych.

* Krzywa Precision-Recall pokazuje wyraźny kompromis między precision a recall — wraz ze zwiększaniem recall precyzja zwykle spada, ponieważ model wykrywa więcej przypadków pozytywnych, ale jednocześnie generuje więcej fałszywych alarmów.

* Najwyższe wartości precision, około 0.65–0.70, występują przy stosunkowo wysokich progach, np. około 0.50, ale wtedy recall jest niski, około 0.15–0.25. Oznacza to, że model wskazuje mniej przypadków jako pozytywne, ale robi to z większą pewnością.

* Dla progów około 0.35–0.45 model osiąga precision mniej więcej w zakresie 0.45–0.60, przy recall około 0.30–0.45. Jest to zakres bardziej ostrożnej klasyfikacji, w którym model ogranicza liczbę fałszywie pozytywnych wskazań, ale pomija część rzeczywistych przypadków pozytywnych.

* Przy niższych progach, około 0.20–0.30, recall rośnie do około 0.65–1.00, ale precision spada do około 0.30–0.45. Oznacza to, że model wykrywa większą część klasy pozytywnej, ale znacznie częściej myli się, klasyfikując obserwacje negatywne jako pozytywne.

**Wnioski**:

* Jeżeli najważniejsze jest maksymalne wykrywanie przypadków pozytywnych, czyli wysoki recall, można rozważyć niski próg decyzyjny, około 0.20–0.25, akceptując precision na poziomie około 0.30–0.35.

* Jeżeli ważniejsza jest większa pewność wykryć, czyli wyższy precision, można rozważyć próg około 0.45–0.50, gdzie precision dochodzi do około 0.60–0.70, ale recall jest niski, mniej więcej 0.20–0.30.

* Zakres pośredni, np. próg około 0.30–0.35, może być kompromisem: recall jest jeszcze umiarkowany, a precision nie spada tak mocno jak przy bardzo niskich progach.

* Najlepszy próg decyzyjny zależy od relatywnego kosztu błędu fałszywie pozytywnego i fałszywie negatywnego. W tym przypadku obniżenie progu zwiększa wykrywalność klasy pozytywnej, ale wyraźnie obniża precyzję klasyfikacji.

Tworzy krzywą Precision-Recall na podstawie wartości rzeczywistych oraz wcześniej obliczonych wyników modelu, np. prawdopodobieństw klasy pozytywnej lub wartości funkcji decyzyjnej. ## DET Curve – Detection Error Tradeoff Curve

*Krzywa kompromisu między błędami detekcji*

**DET Curve** służy do oceny klasyfikatora binarnego, podobnie jak krzywa ROC, ale prezentuje dane w sposób, który **bardziej uwidacznia różnice w jakości klasyfikacji**, zwłaszcza przy niskich poziomach błędów.

---

**Osie wykresu DET**:

* **Oś X** – *False Positive Rate (FPR)*
* **Oś Y** – *False Negative Rate (FNR)*

- Obie osie są zwykle przekształcone przez odwrotną funkcję dystrybuanty normalnej (tzw. **probit scale**).
- Skala jest **nieliniowa** i wyraźniej rozciąga wartości w zakresie niskich błędów.

---

**Definicje**:

* **FPR (False Positive Rate)** – odsetek przypadków negatywnych błędnie zakwalifikowanych jako pozytywne:

  $$  \text{FPR} = \frac{\text{FP}}{\text{FP} + \text{TN}}  $$

* **FNR (False Negative Rate)** – odsetek przypadków pozytywnych błędnie zakwalifikowanych jako negatywne:

  $$  \text{FNR} = \frac{\text{FN}}{\text{FN} + \text{TP}} = 1 - \text{Recall}  $$

---

**Tworzenie krzywej DET:**

1. Oblicz FPR i FNR dla wielu progów decyzyjnych.
2. Przekształć wartości FPR i FNR funkcją probit (odwrotna dystrybuanta normalna).
3. Wykreśl FNR względem FPR.

---

**Interpretacja**:

* Lewy dolny róg wykresu, gdzie FPR ≈ 0 i FNR ≈ 0, odpowiada modelowi idealnemu.
* Im bliżej krzywa znajduje się lewego dolnego rogu, tym lepszy model, ponieważ jednocześnie ogranicza liczbę błędów fałszywie pozytywnych i fałszywie negatywnych.
* Linia FPR=FNR jest użyteczna przy wyznaczaniu punktu Equal Error Rate, czyli progu, dla którego oba typy błędów mają taką samą wartość.

---

**Miara jakości: Equal Error Rate (EER)**

* Punkt przecięcia: **FPR = FNR**

* Wartość EER jest **jedną liczbą charakteryzującą model**:

  $$  \text{EER} = \text{FPR} = \text{FNR} \text{ w punkcie przecięcia}  $$

* Im niższe EER, tym lepszy model.

---

**Zastosowania**:

DET Curve jest preferowana w dziedzinach, gdzie **niskie poziomy błędów są kluczowe**, np.:

* rozpoznawanie mowy,
* biometryka (odcisk palca, twarz, głos),
* systemy bezpieczeństwa,
* systemy detekcji oszustw.

---

**Zalety DET Curve**:

* Lepsze uwidocznienie różnic jakości klasyfikatorów przy **małych wartościach FPR/FNR** niż ROC.
* Umożliwia ocenę działania modeli w rygorystycznych warunkach (np. tolerancja tylko 1% błędów).
* Ułatwia porównanie modeli, które w ROC wyglądają podobnie.

## Confusion Matrix (Macierz Pomyłek)

Macierz pomyłek to podstawowe narzędzie do oceny skuteczności klasyfikatora binarnego. Przedstawia **liczebność przypadków** dla czterech możliwych wyników klasyfikacji.

---

**Struktura macierzy (dla klasyfikacji binarnej)**:

|                              | **Rzeczywista klasa: Pozytywna (1)** | **Rzeczywista klasa: Negatywna (0)** |
| ---------------------------- | ------------------------------------ | ------------------------------------ |
| **Predykcja: Pozytywna (1)** | TP – True Positives                  | FP – False Positives                 |
| **Predykcja: Negatywna (0)** | FN – False Negatives                 | TN – True Negatives                  |

---

**Definicje elementów**:

* **TP (True Positives)** – prawdziwie pozytywne: model poprawnie przewidział klasę pozytywną.
* **FP (False Positives)** – fałszywie pozytywne: model błędnie przypisał klasę pozytywną (tzw. *false alarm*).
* **TN (True Negatives)** – prawdziwie negatywne: model poprawnie przewidział klasę negatywną.
* **FN (False Negatives)** – fałszywie negatywne: model błędnie zakwalifikował przypadek pozytywny jako negatywny.

---

**Metryki wyliczane z macierzy**:

* **Accuracy**: ogólna trafność

  $$  \text{Accuracy} = \frac{TP + TN}{TP + TN + FP + FN}  $$

* **Precision** (Precyzja):

  $$  \text{Precision} = \frac{TP}{TP + FP}  $$

* **Recall (Sensitivity / TPR)**:

  $$  \text{Recall} = \frac{TP}{TP + FN}  $$

* **Specificity (TNR)**:

  $$  \text{Specificity} = \frac{TN}{TN + FP}  $$

* **F1-score** (harmoniczna średnia Precision i Recall):

  $$  \text{F1} = 2 \cdot \frac{\text{Precision} \cdot \text{Recall}}{\text{Precision} + \text{Recall}}  $$

---

**Interpretacja i zastosowanie**:

* **CM uwzględnia konkretny próg decyzyjny** — w przeciwieństwie do ROC/PR, które pokazują działanie modelu dla wszystkich progów.
* Pozwala zobaczyć **strukturę błędów**: np. czy model częściej myli pozytywne z negatywnymi, czy odwrotnie.
* Idealna macierz ma **0** w komórkach FP i FN.

---

**Zalety**:

* Jasne pokazanie, **gdzie model się myli**.
* Konieczne przy ocenie kosztów błędów: np. FN może być gorszy niż FP (np. w diagnozie medycznej).

## Calibration Curve (Krzywa kalibracji, wykres kalibracji)

Wykres kalibracji służy do oceny **zgodności między przewidywanymi prawdopodobieństwami a rzeczywistymi częstościami zdarzeń**. Jest to kluczowe narzędzie wtedy, gdy model zwraca nie tylko końcowe klasy, ale również **prawdopodobieństwa**. Ma szczególne znaczenie w zastosowaniach decyzyjnych, np. w medycynie, finansach, ubezpieczeniach czy systemach oceny ryzyka.

---

**Osie wykresu Calibration Curve**

**Oś X** – *przewidywane prawdopodobieństwo*

Średnia wartość `y_proba` w danym przedziale prawdopodobieństwa, np. 0.1–0.2, 0.2–0.3 itd.

**Oś Y** – *rzeczywista częstość pozytywnych przypadków*

Odsetek obserwacji z `y_true = 1` w danym przedziale przewidywanego prawdopodobieństwa.

---

**Jak powstaje krzywa kalibracji**

1. Obserwacje są dzielone na przedziały według przewidywanego prawdopodobieństwa klasy pozytywnej `y_proba`.

2. W każdym przedziale obliczana jest:

   * średnia wartość `y_proba`,
   * rzeczywista częstość występowania klasy pozytywnej.

3. Następnie wykreślane są punkty:

$$
(\text{średnie przewidywane prawdopodobieństwo},\ \text{rzeczywista częstość klasy pozytywnej})
$$

i łączone linią.

---

**Idealna kalibracja**

Linia:

$$
y = x
$$

oznacza model idealnie skalibrowany.

Przykładowo, jeśli model przypisuje grupie obserwacji prawdopodobieństwo około `0.7`, to w dobrze skalibrowanym modelu około **70% tych obserwacji** powinno rzeczywiście należeć do klasy pozytywnej.

---

**Interpretacja**

* **Krzywa blisko przekątnej** oznacza, że przewidywane prawdopodobieństwa dobrze odpowiadają rzeczywistym częstościom klasy pozytywnej.

* **Krzywa nad przekątną** oznacza, że rzeczywista częstość klasy pozytywnej jest większa niż przewidywane prawdopodobieństwo. Model **zaniża prawdopodobieństwa klasy pozytywnej**.

* **Krzywa pod przekątną** oznacza, że rzeczywista częstość klasy pozytywnej jest mniejsza niż przewidywane prawdopodobieństwo. Model **zawyża prawdopodobieństwa klasy pozytywnej**.

* **Duże odchylenie od przekątnej** oznacza, że prawdopodobieństwa zwracane przez model nie są dobrze skalibrowane. Model może nadal dobrze klasyfikować lub dobrze porządkować obserwacje, ale wartości `predict_proba` należy ostrożnie traktować jako rzeczywiste prawdopodobieństwa.

---

**Zastosowanie**

Wykres kalibracji jest szczególnie przydatny wtedy, gdy ważne jest nie tylko to, **jaką klasę przewiduje model**, ale również **z jakim prawdopodobieństwem**.

Może być wykorzystywany do:

* oceny systemów scoringowych,
* analizy wiarygodności przewidywanych prawdopodobieństw,
* wspomagania wyboru progów decyzyjnych,
* oceny modeli stosowanych do szacowania ryzyka,
* sprawdzenia, czy model wymaga kalibracji.

---

**Kalibracja modelu**

Jeżeli model zwraca prawdopodobieństwa słabo zgodne z rzeczywistymi częstościami zdarzeń, można zastosować metody kalibracji, takie jak:

* `CalibratedClassifierCV`,
* Platt Scaling,
* Isotonic Regression.

Celem kalibracji jest poprawa jakości przewidywanych prawdopodobieństw, bez konieczności zmiany samego modelu bazowego.

---

**Wskazówki praktyczne**

* Histogram lub wykres słupkowy często dodaje się pod krzywą kalibracyjną, aby pokazać, ile obserwacji znajduje się w poszczególnych przedziałach prawdopodobieństwa.

* Modele takie jak Random Forest, Gradient Boosting czy XGBoost mogą osiągać wysoką skuteczność klasyfikacji, ale ich prawdopodobieństwa nie zawsze są dobrze skalibrowane.

* Model może mieć wysokie ROC AUC lub dobre wyniki klasyfikacji, a jednocześnie słabą kalibrację.

* Krzywa kalibracyjna ocenia inny aspekt jakości modelu niż ROC Curve, Precision-Recall Curve czy Confusion Matrix.

---

**Podsumowanie**

Wykres kalibracji pozwala sprawdzić, czy przewidywane przez model prawdopodobieństwa odpowiadają rzeczywistym częstościom występowania klasy pozytywnej. Jest szczególnie ważny wtedy, gdy wyniki modelu są interpretowane probabilistycznie, np. jako ryzyko, szansa wystąpienia zdarzenia lub podstawa do decyzji progowej.


Wykres kalibracji służy do oceny **zgodności między przewidywanymi prawdopodobieństwami a rzeczywistymi częstościami zdarzeń**. To kluczowe narzędzie, gdy model zwraca nie tylko klasy, ale **prawdopodobieństwa** — szczególnie ważne w zastosowaniach decyzyjnych (np. medycyna, finanse, ubezpieczenia).

---

**Osie wykresu Calibration Curve**

* **Oś X** – *przewidywane prawdopodobieństwo*
  Przeciętna wartość `y_proba` w danym przedziale (np. 0.1–0.2, 0.2–0.3, ...)

* **Oś Y** – *rzeczywista częstość pozytywnych przypadków*
  Odsetek przykładów z `y_true = 1` w danym przedziale prawdopodobieństwa

---

**Jak powstaje krzywa kalibracji**:

1. Przypadki są **dzielone na przedziały** według przewidywanego `y_proba` (np. 10 równych przedziałów).
2. W każdym przedziale obliczana jest:

   * **średnia wartość `y_proba`** (X),
   * **częstość rzeczywistych klas pozytywnych** (Y).
     
3. Wykreślamy punkty **(X, Y)** i łączymy linią.

---

**Idealna kalibracja**:

* Linia **y = x** oznacza model idealnie skalibrowany:
  np. jeśli model przypisuje 0.7, to rzeczywiście 70% takich przypadków należy do klasy pozytywnej.

---

**Interpretacja**:

* **Krzywa nad przekątną**: model **zaniża prawdopodobieństwa** – jest **optymistyczny**.
* **Krzywa pod przekątną**: model **zawyża prawdopodobieństwa** – jest **pesymistyczny**.
* **Duże odchylenie od przekątnej** oznacza, że model **nie powinien być używany do decyzji probabilistycznych** (np. ranking, scoring, ocena ryzyka).

---

**Zastosowanie**:

* Kiedy ważne jest nie tylko *kto wygra*, ale *z jakim prawdopodobieństwem*.
* Przydatny do:

  * oceny systemów scoringowych,
  * określenia sensownych progów decyzyjnych,
  * **kalibracji modelu**: np. `CalibratedClassifierCV`, `Platt Scaling`, `Isotonic Regression`.

---

**Wskazówki praktyczne**:

* Histogram (barplot) często dodaje się poniżej krzywej – pokazuje **ile przypadków** model klasyfikował w każdym przedziale.
* Modele typu **Random Forest** lub **XGBoost** często są **źle skalibrowane**, mimo dobrej jakości klasyfikacji (np. wysokie AUC).
* Model może być świetny na ROC Curve, a nadal źle skalibrowany — **to różne aspekty oceny**.

## Lift Chart - Krzywa lift

Krzywa lift (ang. *lift curve*) jest narzędziem służącym do oceny jakości klasyfikatorów, zwłaszcza w zadaniach klasyfikacji binarnej z niezbalansowanymi danymi (np. fraud detection, churn prediction). Jest szczególnie użyteczna, gdy interesuje nas skuteczność modelu w identyfikacji pozytywnej klasy wśród najbardziej prawdopodobnych przypadków.

Lift porównuje skuteczność modelu z losowym wyborem. Pokazuje, ile razy większa jest szansa na wykrycie pozytywnego przypadku, gdy wybieramy próbki zgodnie z rankingiem modelu (np. posortowane według prawdopodobieństwa klasy pozytywnej), w porównaniu z przypadkowym wyborem.

**Krok po kroku**:

1. Model przypisuje każdej obserwacji prawdopodobieństwo przynależności do klasy pozytywnej.
2. Obserwacje są sortowane malejąco według tego prawdopodobieństwa.
3. Dzielimy dane na decyle (lub inne równe grupy, np. 10%, 20%, …).
4. Dla każdej grupy obliczamy:

   * udział obserwacji z klasy pozytywnej,
   * lift = (udział w grupie) / (średni udział klasy pozytywnej w całym zbiorze).

**Interpretacja**:

* **Lift = 1** – model nie lepszy niż losowy wybór.
* **Lift > 1** – model lepszy niż losowy (im wyższy, tym lepszy).
* **Lift < 1** – model gorszy niż losowy (rzadkie i niepożądane).

Krzywa lift może być przedstawiona w dwóch wariantach. **Lift decylowy** pokazuje jakość modelu w każdym przedziale rankingu osobno, np. w najlepszych 10%, kolejnych 10% itd. Pozwala sprawdzić, czy model koncentruje przypadki pozytywne na początku uporządkowanej listy. **Lift skumulowany** pokazuje wynik narastająco, np. dla najlepszych 10%, 20%, 30% obserwacji. Jest szczególnie przydatny, gdy chcemy ocenić, jaki efekt uzyskamy, wybierając określony procent obserwacji z najwyższym prawdopodobieństwem klasy pozytywnej.


```python
df = pd.DataFrame({
    "y_true": y_test,
    "y_proba": y_proba
})
```

```python
def plot_lift_curve(df, n_bins=10, pos_label=1, cumulative=False):

    # cumulative=False pokazuje lift osobno dla każdego decyla
    # cumulative=True pokazuje lift narastająco dla top 10%, top 20%, top 30% itd.
    
    df = df.copy()
    
    # Zamiana klasy pozytywnej na 1, pozostałych na 0
    df["y_positive"] = (df["y_true"] == pos_label).astype(int)
    
    # Sortowanie od najwyższego prawdopodobieństwa klasy pozytywnej
    df = df.sort_values("y_proba", ascending=False).reset_index(drop=True)
    
    # Podział na decyle
    df["bucket"] = pd.qcut(df.index, q=n_bins, labels=False)
    
    baseline = df["y_positive"].mean()
    
    if baseline == 0:
        raise ValueError("Brak klasy pozytywnej. Sprawdź wartość argumentu pos_label.")
    
    lift = []
    thresholds = []
    
    for i in range(n_bins):
        
        if cumulative:
            # Skumulowany lift: od najlepszego decyla do bieżącego
            group = df[df["bucket"] <= i]
        else:
            # Lift decylowy: tylko bieżący decyl
            group = df[df["bucket"] == i]
        
        lift_i = group["y_positive"].mean() / baseline
        lift.append(lift_i)
        
        thresholds.append(group["y_proba"].min())
    
    fig, ax = plt.subplots(figsize=(8, 5))
    
    ax.plot(
        range(1, n_bins + 1),
        lift,
        marker="o",
        label="Cumulative Lift" if cumulative else "Lift"
    )
    
    ax.axhline(
        1.0,
        color="gray",
        linestyle="--",
        label="Random model"
    )
    
    if cumulative:
        ax.set_title("Cumulative Lift Curve")
        ax.set_xlabel("Top deciles included")
        ax.set_ylabel("Cumulative Lift")
    else:
        ax.set_title("Lift Curve")
        ax.set_xlabel("Decile (1 = top 10% highest scores)")
        ax.set_ylabel("Lift")
    
    ax.grid(True, linestyle="--", alpha=0.6)
    ax.legend()
    
    # Dodanie progów prawdopodobieństwa
    for i, threshold in enumerate(thresholds):
        ax.annotate(
            f"{threshold:.2f}",
            (i + 1, lift[i]),
            textcoords="offset points",
            xytext=(0, -12),
            ha="center",
            fontsize=8
        )
    
    plt.tight_layout()
    plt.show()
```

```python
# Lift decylowy
plot_lift_curve(df, n_bins=10, pos_label=1, cumulative=False)
```

```python
# Lift skumulowany
plot_lift_curve(df, n_bins=10, pos_label=1, cumulative=True)
```

**Interpretacja wyników:**

Na wykresie **Lift Curve** widać, że model najlepiej koncentruje klasę pozytywną w pierwszym decylu, czyli wśród **10% obserwacji z najwyższym przewidywanym prawdopodobieństwem klasy pozytywnej**. Lift w pierwszym decylu wynosi około **2.3**, co oznacza, że w tej grupie udział przypadków pozytywnych jest ponad **2 razy większy** niż średnio w całym zbiorze testowym.

Drugi decyl również jest lepszy od wyboru losowego, ponieważ lift wynosi około **1.3**. Oznacza to, że model nadal dość dobrze wskazuje obserwacje pozytywne, choć słabiej niż w pierwszym decylu.

W kolejnych decylach wynik jest mniej stabilny. Trzeci decyl ma lift poniżej **1**, czyli zawiera mniej przypadków pozytywnych niż przeciętnie. Czwarty i piąty decyl ponownie mają lift powyżej **1**, co wskazuje na pewną nieregularność rankingu. Od szóstego decyla lift jest przeważnie niski, co oznacza, że w dolnej części rankingu model znajduje już znacznie mniej przypadków pozytywnych.

Opisy przy punktach, np. **0.50**, **0.40**, **0.34**, oznaczają przybliżone progi prawdopodobieństwa `y_proba`, od których zaczynają się dane decyle. Pierwszy decyl obejmuje więc obserwacje z najwyższymi wynikami, mniej więcej od progu około **0.50**.

**Interpretacja skumulowanej krzywej lift:**

Na wykresie **Cumulative Lift Curve** widać, że jeśli wybierzemy tylko najlepsze **10% obserwacji**, model daje lift około **2.3**. Oznacza to bardzo dobrą koncentrację klasy pozytywnej na początku rankingu.

Dla najlepszych **20% obserwacji** skumulowany lift spada do około **1.8**, a dla najlepszych **30–50%** utrzymuje się w zakresie około **1.5**. To oznacza, że model nadal daje wyraźną korzyść względem losowego wyboru.

Po uwzględnieniu coraz większej części populacji skumulowany lift stopniowo zbliża się do **1.0**. Jest to oczekiwane, ponieważ przy uwzględnieniu **100% obserwacji** wynik zawsze odpowiada średniej w całym zbiorze, więc lift musi wynosić około **1**.

**Wnioski końcowe:**

Model dobrze nadaje się do **rankingowania obserwacji** i wskazywania grupy o najwyższym prawdopodobieństwie klasy pozytywnej. Największa wartość praktyczna jest w pierwszych decylach, szczególnie w top **10–20%** obserwacji.

Jeżeli celem jest wybranie niewielkiej grupy najbardziej ryzykownych lub najbardziej prawdopodobnych przypadków pozytywnych, model daje zauważalną przewagę nad wyborem losowym. Natomiast przy rozszerzaniu analizy na większą część populacji przewaga modelu stopniowo maleje.

Najbardziej sensowny zakres działania to prawdopodobnie wybór obserwacji z górnych **10–30% rankingu**, czyli mniej więcej do progów około **0.34–0.50**, zależnie od tego, ile przypadków chcemy objąć dalszą analizą.

## Krzywa skumulowanego zysku (Cumulative Gain Curve)

**Cumulative Gain Curve** (*krzywa skumulowanego zysku*) to narzędzie oceny jakości klasyfikatorów binarnych, podobne do krzywej lift. Zamiast względnej poprawy względem wyboru losowego pokazuje jednak bezpośrednio, **jaki skumulowany odsetek wszystkich rzeczywistych przypadków pozytywnych udało się uchwycić**, przeglądając coraz większą część próbek posortowanych malejąco według przewidywanego prawdopodobieństwa klasy pozytywnej.

---

**Jak działa:**

1. Model przypisuje każdej obserwacji prawdopodobieństwo przynależności do klasy pozytywnej.
2. Obserwacje są sortowane malejąco według tego prawdopodobieństwa.
3. Oblicza się, jaki procent wszystkich rzeczywistych przypadków pozytywnych znajduje się w kolejnych porcjach danych, np. top 10%, 20%, …, 100%.
4. Wykres przedstawia:

   * **Oś X**: odsetek przeanalizowanych danych, np. top 10%, 20%, 30% obserwacji.
   * **Oś Y**: skumulowany odsetek wszystkich rzeczywistych przypadków pozytywnych uchwyconych w danej części rankingu.

---

**Interpretacja:**

* **Linia przekątna** (*baseline*) oznacza model losowy, np. 20% danych pozwala uchwycić około 20% przypadków pozytywnych.
* **Krzywa powyżej linii przekątnej** oznacza, że model porządkuje obserwacje lepiej niż losowo.
* Im szybciej krzywa rośnie na początku, tym lepszy model, ponieważ szybciej koncentruje przypadki pozytywne na początku rankingu.

---

**Przykład:**

Jeśli model w pierwszych 20% obserwacji, czyli w grupie z najwyższym przewidywanym prawdopodobieństwem klasy pozytywnej, identyfikuje 70% wszystkich rzeczywistych przypadków pozytywnych, oznacza to bardzo skuteczne sortowanie obserwacji według ryzyka lub prawdopodobieństwa klasy pozytywnej.

```python
def compute_cumulative_gain(
    model,
    X_train,
    y_train,
    X_test,
    y_test,
    n_bins=10,
    pos_label=1
):
    # Trenowanie modelu
    model.fit(X_train, y_train)
    
    # Wynik modelu: prawdopodobieństwo klasy pozytywnej albo przeskalowana decision_function
    if hasattr(model, "predict_proba"):
        positive_class_index = list(model.classes_).index(pos_label)
        y_score = model.predict_proba(X_test)[:, positive_class_index]
    else:
        y_score = model.decision_function(X_test)
        y_score = (y_score - y_score.min()) / (y_score.max() - y_score.min())
    
    # Przygotowanie danych
    df = pd.DataFrame({
        "y_true": y_test,
        "y_score": y_score
    })
    
    # Zamiana klasy pozytywnej na 1, pozostałych na 0
    df["y_positive"] = (df["y_true"] == pos_label).astype(int)
    
    # Sortowanie od najwyższego prawdopodobieństwa klasy pozytywnej
    df = df.sort_values("y_score", ascending=False).reset_index(drop=True)
    
    # Podział na decyle
    df["bucket"] = pd.qcut(df.index, q=n_bins, labels=False)
    
    total_positives = df["y_positive"].sum()
    
    if total_positives == 0:
        raise ValueError("Brak klasy pozytywnej w y_test. Sprawdź wartość pos_label.")
    
    gain = []
    thresholds = []
    cum_tp = 0
    
    grouped = df.groupby("bucket")
    
    for i in range(n_bins):
        group = grouped.get_group(i)
        
        cum_tp += group["y_positive"].sum()
        gain_i = cum_tp / total_positives
        gain.append(gain_i)
        
        thresholds.append(group["y_score"].min())
    
    results = pd.DataFrame({
        "bucket": np.arange(1, n_bins + 1),
        "sample_fraction": np.arange(1, n_bins + 1) / n_bins,
        "cumulative_gain": gain,
        "threshold": thresholds
    })
    
    return results
```

```python
gain_df = compute_cumulative_gain(
    model,
    X_train,
    y_train,
    X_test,
    y_test,
    n_bins=10,
    pos_label=1
)

gain_df
```

```python
fig, ax = plt.subplots(figsize=(8, 5))

ax.plot(
    gain_df["sample_fraction"],
    gain_df["cumulative_gain"],
    marker="o",
    label="Model"
)

ax.plot(
    [0, 1],
    [0, 1],
    linestyle="--",
    color="gray",
    label="Random model"
)

for _, row in gain_df.iterrows():
    ax.annotate(
        f"{row['threshold']:.2f}",
        (row["sample_fraction"], row["cumulative_gain"]),
        textcoords="offset points",
        xytext=(0, -12),
        ha="center",
        fontsize=8
    )

ax.set_xlabel("Fraction of sample")
ax.set_ylabel("Cumulative gain")
ax.set_title("Cumulative Gain Curve")
ax.grid(True, linestyle="--", alpha=0.6)
ax.legend()

plt.tight_layout()
plt.show()
```

**Interpretacja:**

Krzywa **Cumulative Gain Curve** znajduje się wyraźnie powyżej linii modelu losowego, co oznacza, że model porządkuje obserwacje lepiej niż przypadkowy wybór.

W pierwszych **10% obserwacji** z najwyższym przewidywanym prawdopodobieństwem klasy pozytywnej model uchwycił około **23% wszystkich rzeczywistych przypadków pozytywnych**. Oznacza to, że już niewielka grupa najwyżej ocenionych obserwacji zawiera ponad dwukrotnie więcej przypadków pozytywnych, niż oczekiwalibyśmy przy wyborze losowym.

Dla najlepszych **20% obserwacji** model uchwycił około **36–37%** wszystkich przypadków pozytywnych, a dla najlepszych **40% obserwacji** około **60%** przypadków pozytywnych. To wskazuje, że model dobrze koncentruje klasę pozytywną w górnej części rankingu.

Przy wyborze najlepszych **50% obserwacji** model obejmuje już około **73–74%** wszystkich przypadków pozytywnych. Po tym punkcie krzywa nadal rośnie, ale przewaga nad modelem losowym stopniowo maleje.

Opisy przy punktach, np. **0.50**, **0.40**, **0.34**, oznaczają minimalne wartości przewidywanego prawdopodobieństwa klasy pozytywnej w danym zakresie rankingu. Na przykład próg około **0.30** odpowiada mniej więcej wyborowi najlepszych **40% obserwacji**.

**Wnioski:**

Model jest użyteczny do rankingowania obserwacji według prawdopodobieństwa klasy pozytywnej. Największa korzyść pojawia się w początkowej części rankingu, szczególnie dla top **10–50%** obserwacji.

Jeżeli celem jest szybkie wykrycie dużej części przypadków pozytywnych przy ograniczonej liczbie obserwacji do dalszej analizy, sensownym zakresem może być wybór top **30–50%** rankingu. W tym zakresie model obejmuje około **45–74%** wszystkich przypadków pozytywnych i nadal wyraźnie przewyższa wybór losowy.

Przy analizie całego zbioru, czyli **100% obserwacji**, krzywa musi dojść do wartości **1.0**, ponieważ obejmujemy wtedy wszystkie przypadki pozytywne. Dlatego największe znaczenie praktyczne ma kształt krzywej na początku, a nie jej końcowy punkt.

## CAP Curve - Cumulative Accuracy Profile Curve

CAP Curve pokazuje, jaki odsetek wszystkich rzeczywistych obserwacji należących do klasy pozytywnej można wykryć, analizując kolejne części zbioru uporządkowane malejąco według przewidywanego prawdopodobieństwa klasy pozytywnej.

Jeśli klasa pozytywna to 1, to CAP Curve pokazuje, jaki odsetek wszystkich rzeczywistych przypadków 1 uda się uchwycić, jeśli będziemy analizować np. top 10%, 20%, 30% obserwacji z najwyższym y_proba.

CAP Curve jest bardzo podobna do **Cumulative Gain Curve**. Różnica polega głównie na sposobie interpretacji i częstym dodaniu linii modelu idealnego. Wbrew nazwie, CAP Curve nie odnosi się bezpośrednio do metryki accuracy, tylko do skumulowanego wykrywania klasy pozytywnej.

```python
df = pd.DataFrame({
    "y_true": y_test,
    "y_proba": y_proba
})
```

```python
def plot_cap_curve(df, n_bins=10, pos_label=1):
    df = df.copy()
    
    # Zamiana klasy pozytywnej na 1, pozostałych na 0
    df["y_positive"] = (df["y_true"] == pos_label).astype(int)
    
    # Sortowanie od najwyższego prawdopodobieństwa klasy pozytywnej
    df = df.sort_values("y_proba", ascending=False).reset_index(drop=True)
    
    total_observations = len(df)
    total_positives = df["y_positive"].sum()
    
    if total_positives == 0:
        raise ValueError("Brak klasy pozytywnej. Sprawdź wartość argumentu pos_label.")
    
    # Podział na decyle
    df["bucket"] = pd.qcut(df.index, q=n_bins, labels=False)
    
    grouped = df.groupby("bucket")
    
    sample_fraction = []
    cap_values = []
    thresholds = []
    
    cumulative_observations = 0
    cumulative_positives = 0
    
    for i in range(n_bins):
        group = grouped.get_group(i)
        
        cumulative_observations += len(group)
        cumulative_positives += group["y_positive"].sum()
        
        sample_fraction.append(cumulative_observations / total_observations)
        cap_values.append(cumulative_positives / total_positives)
        thresholds.append(group["y_proba"].min())
    
    # Dodanie punktu początkowego
    sample_fraction = [0] + sample_fraction
    cap_values = [0] + cap_values
    
    # Model idealny
    positive_rate = total_positives / total_observations
    perfect_x = [0, positive_rate, 1]
    perfect_y = [0, 1, 1]
    
    fig, ax = plt.subplots(figsize=(8, 5))
    
    ax.plot(
        sample_fraction,
        cap_values,
        marker="o",
        label="Model"
    )
    
    ax.plot(
        [0, 1],
        [0, 1],
        linestyle="--",
        color="gray",
        label="Random model"
    )
    
    ax.plot(
        perfect_x,
        perfect_y,
        linestyle=":",
        color="black",
        label="Perfect model"
    )
    
    # Opisy progów decyzyjnych przy punktach
    for i, threshold in enumerate(thresholds):
        ax.annotate(
            f"{threshold:.2f}",
            (sample_fraction[i + 1], cap_values[i + 1]),
            textcoords="offset points",
            xytext=(0, -12),
            ha="center",
            fontsize=8
        )
    
    ax.set_title("CAP Curve")
    ax.set_xlabel("Fraction of sample")
    ax.set_ylabel("Fraction of positives captured")
    ax.grid(True, linestyle="--", alpha=0.6)
    ax.legend()
    
    plt.tight_layout()
    plt.show()
```

```python
plot_cap_curve(df, n_bins=10, pos_label=1)
```

Na wykresie:

* linia Random model oznacza wybór losowy,
* linia Perfect model oznacza model idealny,
* krzywa Model pokazuje, jak dobrze Twój model koncentruje przypadki "bad" na początku rankingu,
* wartości przy punktach oznaczają minimalne y_proba w danym zakresie rankingu.

Linia modelu idealnego w CAP Curve pokazuje sytuację, w której model umieszcza wszystkie przypadki pozytywne na samym początku rankingu.

Interpretacja CAP Curve:

Krzywa modelu znajduje się wyraźnie powyżej linii modelu losowego, co oznacza, że model porządkuje obserwacje lepiej niż przypadkowy wybór. Przypadki klasy pozytywnej są częściej skoncentrowane na początku rankingu, czyli wśród obserwacji z najwyższym przewidywanym prawdopodobieństwem klasy pozytywnej.

W pierwszych 10% obserwacji model uchwycił około 23% wszystkich przypadków pozytywnych. Jest to wynik lepszy niż losowy, ponieważ model losowy przy 10% obserwacji uchwyciłby średnio tylko około 10% przypadków pozytywnych.

Dla najlepszych 20% obserwacji model uchwycił około 36–37% przypadków pozytywnych, a dla najlepszych 40% obserwacji około 60% wszystkich przypadków pozytywnych. Oznacza to, że model dość dobrze koncentruje przypadki pozytywne w górnej części rankingu.

Dla najlepszych 50–60% obserwacji model obejmuje już około 73–75% przypadków pozytywnych. Powyżej tego zakresu krzywa nadal rośnie, ale zysk z dodawania kolejnych obserwacji jest coraz mniejszy.

Wartości przy punktach, np. 0.50, 0.40, 0.34, oznaczają przybliżone progi y_proba, czyli minimalne przewidywane prawdopodobieństwo klasy pozytywnej w danym zakresie rankingu. Na przykład próg około 0.30 odpowiada analizie około 40% obserwacji z najwyższym wynikiem modelu.

Wnioski:

Model ma użyteczną zdolność rankingowania obserwacji, ponieważ szybciej wykrywa przypadki pozytywne niż wybór losowy. Największa korzyść jest widoczna w pierwszych 40–50% rankingu, gdzie udaje się uchwycić znaczną część przypadków pozytywnych.

Model nie jest jednak bliski modelowi idealnemu, ponieważ krzywa modelu znajduje się wyraźnie poniżej czarnej linii modelu doskonałego. Oznacza to, że przypadki pozytywne nie są całkowicie skoncentrowane na samym początku rankingu.

W praktyce sensowne może być analizowanie obserwacji do progu około 0.30–0.34, czyli górnych 30–40% rankingu, jeśli celem jest uchwycenie dużej części przypadków pozytywnych przy ograniczeniu liczby obserwacji kierowanych do dalszej analizy.
