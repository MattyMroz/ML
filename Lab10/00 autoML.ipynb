# autoML

**AutoML** (automatyczne uczenie maszynowe) to zbiór metod i narzędzi, które automatyzują proces projektowania, trenowania i oceny modeli uczenia maszynowego. Celem AutoML jest umożliwienie skutecznego tworzenia modeli nawet osobom "*bez głębokiej wiedzy specjalistycznej*", a także przyspieszenie pracy analityków danych.

1. **Wstępne przetwarzanie danych**

   * Detekcja brakujących danych i ich imputacja
   * Skalowanie i normalizacja zmiennych
   * Kodowanie zmiennych kategorycznych
   * Usuwanie kolinearności i wybór cech

2. **Inżynieria cech (feature engineering)**

   * Automatyczne tworzenie nowych cech na podstawie istniejących
   * Redukcja wymiarowości
   * Selekcja najbardziej informacyjnych cech

3. **Dobór modelu (model selection)**

   * Przegląd różnych algorytmów (np. drzewa decyzyjne, SVM, sieci neuronowe)
   * Automatyczne dopasowanie najlepszego modelu do problemu klasyfikacji, regresji, itd.

4. **Strojenie hiperparametrów (hyperparameter tuning)**

   * Optymalizacja parametrów modeli z użyciem np. wyszukiwania siatki (grid search), losowego (random search) lub algorytmów bayesowskich

5. **Walidacja i ocena modelu**

   * Automatyczne dobieranie strategii walidacji (np. k-krotna walidacja krzyżowa)
   * Ocena wyników za pomocą metryk (np. accuracy, AUC, RMSE)

6. **Ensemble learning (opcjonalnie)**

   * Tworzenie zespołów modeli w celu zwiększenia dokładności (np. stacking, bagging, boosting)

7. **Eksport modelu i integracja**

   * Generowanie kodu do wdrożenia modelu
   * Umożliwienie łatwej integracji z aplikacjami produkcyjnymi

## Popularne biblioteki AutoML

* **H2O AutoML** - szybkie, skalowalne narzędzie wspierające klasyfikację, regresję i deep learning.
* **AutoGluon** - rozwijana przez Amazon, dobrze radzi sobie z małymi i dużymi zbiorami danych.
* **MLJAR AutoML** - generuje szczegółowe raporty i przejrzyste modele, wspiera klasyfikację, regresję i time-series.

oraz

* **Auto-sklearn** - oparta na scikit-learn, używa optymalizacji bayesowskiej i meta-learningu (wymagany Linux).
* **AutoKeras** - automatyzuje deep learning w oparciu o Keras i TensorFlow (rekomendowany Linux).
* **Microsoft Azure AutoML** - platforma chmurowa do automatycznego doboru modeli i hiperparametrów (z tego samego ekosystemu jest ML.NET, czyli biblioteka open-source do uczenia maszynowego dla .NET).
  
i inne (wyszukane w Iternecie i trudno mi o nich powiedzieć coś więcej 🙂):

* **TPOT** - wykorzystuje algorytmy genetyczne do automatycznej optymalizacji pipelineów ML.
* **Google Cloud AutoML** - rozwiązanie chmurowe z interfejsem GUI, wspiera NLP, wizję i tabular.
* **FLAML** - bardzo szybka i lekka biblioteka od Microsoft, zoptymalizowana pod kątem efektywności.
* **PyCaret** - wysokopoziomowy interfejs dla wielu typów zadań ML z prostym API.
* **IBM AutoAI** - zintegrowane z IBM Watson Studio, automatyzuje cały pipeline analizy danych.
* **Amazon SageMaker Autopilot** - automatyzuje proces ML w środowisku AWS bez konieczności kodowania.
* **BigML** - narzędzie online z szeroką obsługą algorytmów i wizualizacjami.
* **RapidMiner Auto Model** - graficzne środowisko AutoML dla klasyfikacji i regresji.
* **Optuna** - framework do optymalizacji hiperparametrów, często używany jako komponent AutoML.
