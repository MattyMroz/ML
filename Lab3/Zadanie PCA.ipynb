**Zbiór Wine dataset** zawiera dane chemiczne win wyprodukowanych z trzech różnych szczepów winogron uprawianych w tym samym regionie we Włoszech. Jest wykorzystywany w analizie danych, klasyfikacji oraz do testowania algorytmów uczenia maszynowego.

**Cechy zbioru**:

1. **Alcohol** – zawartość alkoholu (%) w winie.
2. **Malic acid** – zawartość kwasu jabłkowego (g/l), wpływającego na kwasowość i smak wina.
3. **Ash** – pozostałość mineralna po spaleniu próbki (g/l), odzwierciedla zawartość substancji nieorganicznych.
4. **Alcalinity of ash** – zasadowość popiołu, wskaźnik składu mineralnego i stabilności chemicznej.
5. **Magnesium** – zawartość magnezu (mg/l), wpływa na fermentację i stabilność koloru.
6. **Total phenols** – całkowita zawartość fenoli, odpowiedzialnych za smak, cierpkość i trwałość wina.
7. **Flavanoids** – grupa fenoli wpływających na kolor, smak i właściwości zdrowotne.
8. **Nonflavanoid phenols** – fenole nieflawonoidowe, mające wpływ na gorycz.
9. **Proanthocyanins** – związki polifenolowe będące prekursorami antocyjanów, wpływają na barwę i cierpkość.
10. **Color intensity** – intensywność koloru wina; wyższa wartość oznacza ciemniejsze wino.
11. **Hue** – odcień koloru, wskazuje na wiek i typ wina.
12. **OD280/OD315 of diluted wines** – stosunek pochłaniania światła przy długościach fali 280 i 315 nm; miernik jakości fenoli i potencjału starzenia.
13. **Proline** – zawartość prolina (mg/l), aminokwas odzwierciedlający dojrzewanie i jakość winogron.

Zbiór ten zawiera 178 próbek podzielonych na trzy klasy, co pozwala na prowadzenie analiz klasyfikacyjnych i eksploracyjnych.

Oprócz klasycznego **Wine dataset** z UCI istnieje kilka innych, znanych i często wykorzystywanych zbiorów danych dotyczących win. Oto najważniejsze z nich:  

1. **Wine Quality Dataset (UCI)**  
- Dwa zbiory: czerwone wino (*red wine*) i białe wino (*white wine*).  
- Dane pochodzą z portugalskich win z regionu Vinho Verde.  
- Zawiera cechy chemiczne i ocenę jakości sensorycznej w skali od 0 do 10.  
- **Liczba próbek**:  
  - Red wine: 1599 próbek  
  - White wine: 4898 próbek  
- **Cechy obejmują m.in.**: kwasowość, pH, siarczyny, alkohol, cukry resztkowe, gęstość, chlorki.  
- Często używany do problemów regresji lub klasyfikacji jakości.  

2. **Wine Reviews Dataset (Kaggle)**  
- Zawiera ponad 130 000 recenzji win.  
- Kolumny: kraj, region, szczep, producent, ocena punktowa, cena, opis tekstowy.  
- Przydatny do analizy języka naturalnego (NLP), predykcji ocen oraz eksploracji danych geograficznych i cenowych.  

3. **Italian Wine Data (DataHub)**  
- Zbiór obejmuje oceny i cechy win włoskich, wraz z informacjami o regionach, rodzajach szczepów i ocenach sensorycznych.  

4. **Wine Spectator Dataset (dostępny komercyjnie)**  
- Ogromna baza recenzji win z całego świata, używana w badaniach dotyczących analizy tekstu, korelacji cen, ocen i trendów rynkowych.
