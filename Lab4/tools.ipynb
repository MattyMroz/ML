```python
import numpy as np
import matplotlib.pyplot as plt
import matplotlib.patches as patches
from collections import Counter
import numpy as np
```

```python
def draw_index_colorbar(colored_index_groups, colors=None, rect_height=1.0, rect_width=1.0, title=None):
    """
    Rysuje poziomy pasek złożony z prostokątów, z których wybrane są kolorowane według indeksów.

    Parametry:
        colored_index_groups: list[list[int]] – lista grup indeksów (każda grupa będzie miała inny kolor).
        colors: list[str] lub None – lista kolorów (jeśli None, zostaną użyte domyślne).
        rect_height: float – wysokość prostokątów.
        rect_width: float – szerokość każdego prostokąta.
        title: str – opcjonalny tytuł wykresu.
    """
    # Ustal maksymalny indeks (liczba prostokątów = max_index + 1)
    max_index = max([max(group) if len(group) > 0 else 0 for group in colored_index_groups])
    total_length = max_index + 1

    # Ustal kolory (jeśli nie podano)
    if colors is None:
        cmap = plt.get_cmap("tab10")
        colors = [cmap(i) for i in range(len(colored_index_groups))]

    # Tworzenie mapy kolorów
    color_map = {}
    for group, color in zip(colored_index_groups, colors):
        for idx in group:
            color_map[idx] = color

    # Ustawienie rozmiaru figury proporcjonalnie do liczby prostokątów
    fig_width = total_length * rect_width / 10
    fig_height = rect_height * 1.5
    fig, ax = plt.subplots(figsize=(fig_width, fig_height))

    # Rysowanie prostokątów
    for i in range(total_length):
        color = color_map.get(i, 'lightgray')
        rect = patches.Rectangle(
            (i * rect_width, 0), rect_width, rect_height,
            facecolor=color, edgecolor='black', linewidth=0.2
        )
        ax.add_patch(rect)

    ax.set_xlim(0, total_length * rect_width)
    ax.set_ylim(0, rect_height)
    ax.set_aspect('auto')
    ax.axis('off')
    if title:
        ax.set_title(title, pad=10)

    plt.subplots_adjust(top=0.85, bottom=0.1, left=0.01, right=0.99)
    plt.show()
```

```python
def split_indices_by_target(target):
    """
    Dzieli indeksy na grupy według unikalnych wartości targetu (klasy).

    Parametry:
        target: array-like – wektor etykiet klas (np. y z klasyfikacji)

    Zwraca:
        Listę list indeksów, gdzie każda lista odpowiada jednej klasie.
    """
    target = np.array(target)
    unique_classes = np.unique(target)
    index_groups = [(target == cls).nonzero()[0].tolist() for cls in unique_classes]
    return index_groups
```

```python
def count_named_classes_by_indices(target, indices, class_names=None):
    """
    Zlicza wystąpienia klas po indeksach i opcjonalnie zamienia je na nazwy.

    Parametry:
        target: array-like – wektor klas (np. y)
        indices: list[int] – lista indeksów, które mają być zliczane
        class_names: list[str] lub None – nazwy klas w kolejności odpowiadającej etykietom (0, 1, 2, ...)

    Zwraca:
        Słownik {nazwa_klasy: liczba wystąpień} lub {klasa_numeryczna: liczba wystąpień}, jeśli nie podano nazw
    """
    target = np.array(target)
    selected = target[indices]
    counts = Counter(selected)

    if class_names is not None:
        return {class_names[k]: v for k, v in counts.items()}
    else:
        return dict(counts)
```

```python
def count_classes_from_split(y_part, class_names=None):
    """
    Zlicza ilość klas w y_train lub y_test z podziału train_test_split.

    Parametry:
        y_part: array-like – wektor klas z train/test splitu (np. y_train lub y_test)
        class_names: list[str] lub None – opcjonalne etykiety klas

    Zwraca:
        Słownik {klasa/nazwa: liczba wystąpień}
    """
    y_part = np.array(y_part)
    counts = Counter(y_part)

    if class_names is not None:
        return {class_names[k]: v for k, v in counts.items()}
    else:
        return dict(counts)
```
