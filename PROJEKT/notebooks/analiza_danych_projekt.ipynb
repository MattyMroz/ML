### Wykres korelacji - najsilniejsze zależności

```python
# Wybór zmiennych numerycznych
numeric_cols = df.select_dtypes(include=["number"])
corr_matrix = numeric_cols.corr()

# Wyodrębnienie par korelacji
mask = np.triu(np.ones(corr_matrix.shape), k=1).astype(bool)
corr_pairs = (
    corr_matrix.where(mask)
    .stack()
    .reset_index()
    .rename(columns={"level_0": "Zmienna_1", "level_1": "Zmienna_2", 0: "Korelacja"})
)

# Top 10 najwyższych i 10 najniższych korelacji
top_10 = corr_pairs.nlargest(10, "Korelacja")
bottom_10 = corr_pairs.nsmallest(10, "Korelacja")
results = pd.concat([bottom_10, top_10]).sort_values("Korelacja", ascending=True).reset_index(drop=True)

# Tworzenie wykresu
plt.figure(figsize=(14, 10))
pairs_labels = results["Zmienna_1"] + " - " + results["Zmienna_2"]

# Gradient kolorów :)
warm_cmap = LinearSegmentedColormap.from_list("warm", ["#E91E63", "#FF6B6B", "#FF8E53", "#FFB347", "#FF5722"])
norm = plt.Normalize(results["Korelacja"].min(), results["Korelacja"].max())
colors = [warm_cmap(norm(val)) for val in results["Korelacja"]]

bars = plt.barh(pairs_labels, results["Korelacja"], color=colors, edgecolor='white', linewidth=0.5)
plt.xlabel("Współczynnik korelacji Pearsona", fontsize=12)
plt.ylabel("Pary zmiennych", fontsize=12)
plt.title("10 najniższych i 10 najwyższych korelacji w zbiorze danych", fontsize=14, fontweight='bold')
plt.axvline(x=0, color='black', linestyle='-', linewidth=0.5)
plt.grid(axis="x", linestyle="--", alpha=0.5)
plt.tight_layout()
plt.show()
```

**Interpretacja wykresu korelacji:**
- Silna dodatnia korelacja między ocenami z 1. i 2. semestru - studenci utrzymują poziom nauki
- Wiek przy zapisie koreluje z trybem wieczorowym - starsi studenci częściej wybierają tryb wieczorowy

```python
# Kodowanie zmiennych kategorycznych (one-hot encoding)
categorical_cols = [
    "Marital status",
    "Application mode",
    "Course",
    "Daytime/evening attendance",
    "Previous qualification",
    "Nacionality",
    "Mother's qualification",
    "Father's qualification",
    "Mother's occupation",
    "Father's occupation",
    "Application order"
]

df_encoded = pd.get_dummies(df, columns=categorical_cols, prefix=categorical_cols)
print(f"Liczba kolumn po kodowaniu one-hot: {df_encoded.shape[1]}")
```

## b) Wybór cech

Do selekcji najważniejszych cech wykorzystam dwie metody:
1. **Korelacja Pearsona** - mierzy liniową zależność między cechami a zmienną docelową
2. **Test Chi-kwadrat** - ocenia istotność statystyczną cech kategorycznych

```python
# Przygotowanie danych do selekcji cech
X = df_encoded.drop(columns=["Target"])
y = df_encoded["Target"]

# Liczba cech do wybrania
num_features = 30
```

```python
# Selekcja metodą korelacji Pearsona
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)
X_scaled_df = pd.DataFrame(X_scaled, columns=X.columns)

# Obliczenie korelacji z Target
pearson_corr = X_scaled_df.corrwith(y).abs()
pearson_top = pearson_corr.sort_values(ascending=False).head(num_features)

print("Top 30 cech wg korelacji Pearsona:")
print(pearson_top.index.tolist())
```

```python
# Selekcja metodą Chi-kwadrat
# Chi-kwadrat wymaga wartości nieujemnych
X_chi = X.copy()
X_chi["Inflation rate"] = X_chi["Inflation rate"].clip(lower=0)
X_chi["GDP"] = X_chi["GDP"].clip(lower=0)

# Selekcja cech
chi_selector = SelectKBest(score_func=chi2, k=num_features)
chi_selector.fit(X_chi, y)
chi_scores = pd.Series(chi_selector.scores_, index=X.columns)
chi_top = chi_scores.sort_values(ascending=False).head(num_features)

print("\nTop 30 cech wg testu Chi-kwadrat:")
print(chi_top.index.tolist())
```

```python
# Wyznaczenie części wspólnej obu metod selekcji
pearson_set = set(pearson_top.index)
chi_set = set(chi_top.index)
common_features = list(pearson_set & chi_set)

print(f"\nLiczba cech wspólnych: {len(common_features)}")
print("Wybrane cechy:")
print(common_features)

# Zredukowany zbiór danych
X_selected = X[common_features]
```

## c) Przekształcenia danych (normalizacja)

```python
# Normalizacja danych do zakresu [0, 1]
minmax_scaler = MinMaxScaler()
X_normalized = minmax_scaler.fit_transform(X_selected)
X_normalized_df = pd.DataFrame(X_normalized, columns=common_features)

print("Znormalizowane dane - podgląd:")
X_normalized_df.head()
```

# 4. Tworzenie modelu i ocena algorytmów

## a) Podział zbioru danych

Proporcja 70/30 zapewnia wystarczającą ilość danych do treningu i wiarygodną ocenę. Parametr `stratify=y` zachowuje proporcje klas w obu zbiorach, a `random_state=42` gwarantuje powtarzalność wyników.

```python
# Podział na zbiór treningowy (70%) i testowy (30%)
X_train, X_test, y_train, y_test = train_test_split(
    X_normalized_df,
    y,
    test_size=0.30,
    random_state=42,
    stratify=y  # zachowanie proporcji klas
)

print(f"Rozmiar zbioru treningowego: {X_train.shape[0]}")
print(f"Rozmiar zbioru testowego: {X_test.shape[0]}")
```

## b) Algorytmy klasyfikacji

### Model 1: Las Losowy (Random Forest)

**Opis metody:** Las losowy to zespół drzew decyzyjnych, gdzie każde drzewo jest trenowane na losowej próbce danych. Finalna predykcja to wynik głosowania większościowego.

```python
# Etykiety klas do wizualizacji
class_labels = ["Dropout", "Enrolled", "Graduate"]

# Definicja i trening modelu Random Forest
rf_model = RandomForestClassifier(
    n_estimators=100,       # liczba drzew
    max_depth=15,           # maksymalna głębokość
    class_weight="balanced", # balansowanie klas
    random_state=42
)
rf_model.fit(X_train, y_train)

# Predykcja
rf_pred = rf_model.predict(X_test)

# Ocena modelu
rf_accuracy = accuracy_score(y_test, rf_pred)
print(f"Dokładność modelu Random Forest: {rf_accuracy * 100:.2f}%")

# Macierz pomyłek
plt.figure(figsize=(8, 6))
cm_rf = confusion_matrix(y_test, rf_pred)
sns.heatmap(
    cm_rf,
    annot=True,
    fmt="d",
    cmap="Oranges",
    xticklabels=class_labels,
    yticklabels=class_labels,
    linewidths=0.5,
    linecolor='white'
)
plt.title("Macierz pomyłek - Random Forest", fontsize=14, fontweight='bold')
plt.xlabel("Przewidywana klasa", fontsize=12)
plt.ylabel("Rzeczywista klasa", fontsize=12)
plt.tight_layout()
plt.show()
```

### Model 2: Maszyna Wektorów Nośnych (SVM)

**Opis metody:** SVM szuka hiperpłaszczyzny maksymalnie separującej klasy. Użycie jądra liniowego pozwala na efektywną klasyfikację w przestrzeni wielowymiarowej.

```python
# Definicja i trening modelu SVM
svm_model = SVC(
    kernel="linear",    # funkcja jądra
    C=1.0,              # parametr regularyzacji
    random_state=42
)
svm_model.fit(X_train, y_train)

# Predykcja
svm_pred = svm_model.predict(X_test)

# Ocena modelu
svm_accuracy = accuracy_score(y_test, svm_pred)
print(f"Dokładność modelu SVM: {svm_accuracy * 100:.2f}%")

# Macierz pomyłek
plt.figure(figsize=(8, 6))
cm_svm = confusion_matrix(y_test, svm_pred)
sns.heatmap(
    cm_svm,
    annot=True,
    fmt="d",
    cmap="Reds",
    xticklabels=class_labels,
    yticklabels=class_labels,
    linewidths=0.5,
    linecolor='white'
)
plt.title("Macierz pomyłek - SVM", fontsize=14, fontweight='bold')
plt.xlabel("Przewidywana klasa", fontsize=12)
plt.ylabel("Rzeczywista klasa", fontsize=12)
plt.tight_layout()
plt.show()
```

## c) Porównanie wyników algorytmów

```python
# Zestawienie wyników
results_df = pd.DataFrame({
    "Model": ["Random Forest", "SVM"],
    "Dokładność (%)": [rf_accuracy * 100, svm_accuracy * 100]
})
print("\nPorównanie algorytmów:")
print(results_df.to_string(index=False))
```

**Obserwacje:**
- Random Forest lepiej radzi sobie z rozpoznawaniem studentów rezygnujących (Dropout)
- SVM wykazuje większą precyzję w klasyfikacji absolwentów (Graduate)
- Oba modele mają trudności z klasą Enrolled (najmniej liczna klasa)

## d) Przykładowy wynik klasyfikacji

```python
# Przykładowa predykcja dla 5 losowych studentów
np.random.seed(42)
sample_indices = np.random.choice(X_test.index, 5, replace=False)
sample_data = X_normalized_df.loc[sample_indices]
sample_actual = y.loc[sample_indices]
sample_pred = rf_model.predict(sample_data)

# Mapowanie wartości na etykiety
label_map = {0: "Dropout", 1: "Enrolled", 2: "Graduate"}

print("Przykładowe predykcje (Random Forest):")
print("-" * 50)
for idx, (actual, pred) in enumerate(zip(sample_actual.values, sample_pred)):
    status = "[POPRAWNIE]" if actual == pred else "[BŁĘDNIE]"
    print(f"Student {idx+1}: Rzeczywisty={label_map[actual]}, Przewidywany={label_map[pred]} {status}")
```

## e) Czynności poprawiające wynik

Spróbuję poprawić wyniki modelu Random Forest poprzez:
1. Zwiększenie liczby drzew (`n_estimators`)
2. Dostosowanie głębokości drzew (`max_depth`)
3. Usunięcie `class_weight` dla lepszej klasyfikacji klasy Graduate

```python
# Zoptymalizowany model Random Forest
rf_optimized = RandomForestClassifier(
    n_estimators=150,       # więcej drzew
    max_depth=20,           # większa głębokość
    min_samples_split=5,    # min. próbek do podziału
    random_state=42
)
rf_optimized.fit(X_train, y_train)

# Predykcja i ocena
rf_opt_pred = rf_optimized.predict(X_test)
rf_opt_accuracy = accuracy_score(y_test, rf_opt_pred)

print(f"Dokładność przed optymalizacją: {rf_accuracy * 100:.2f}%")
print(f"Dokładność po optymalizacji:    {rf_opt_accuracy * 100:.2f}%")
print(f"Poprawa: {(rf_opt_accuracy - rf_accuracy) * 100:.2f} p.p.")

# Macierz pomyłek po optymalizacji
plt.figure(figsize=(8, 6))
cm_opt = confusion_matrix(y_test, rf_opt_pred)
sns.heatmap(
    cm_opt,
    annot=True,
    fmt="d",
    cmap="YlOrRd",
    xticklabels=class_labels,
    yticklabels=class_labels,
    linewidths=0.5,
    linecolor='white'
)
plt.title("Macierz pomyłek - Random Forest (zoptymalizowany)", fontsize=14, fontweight='bold')
plt.xlabel("Przewidywana klasa", fontsize=12)
plt.ylabel("Rzeczywista klasa", fontsize=12)
plt.tight_layout()
plt.show()
```
